## 1. Revisão: iterando e filtrando com `for`

Na Aula 04 já combinamos `for` com `if`. Vamos aplicar isso para **filtrar** uma lista, ou seja, construir uma nova lista contendo apenas os elementos que atendem a uma condição.

````python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2, 95.5, 60.1]
limite_alerta = 80.0

leituras_em_alerta = []          # lista vazia que vamos preencher
for temperatura in leituras_temperatura:
    if temperatura >= limite_alerta:
        leituras_em_alerta.append(temperatura)

print("Leituras em alerta:", leituras_em_alerta)
````

## 2. *List comprehension*: a mesma filtragem, de forma compacta

Python oferece uma sintaxe mais direta para criar listas a partir de outra sequência: a **compreensão de lista** (*list comprehension*). O padrão é:

```python
[expressao for item in sequencia if condicao]
```

O trecho `if condicao` é opcional.

````python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2, 95.5, 60.1]
limite_alerta = 80.0

# Exatamente o mesmo resultado do laço 'for' acima, em uma linha:
leituras_em_alerta = [temperatura for temperatura in leituras_temperatura if temperatura >= limite_alerta]

print("Leituras em alerta:", leituras_em_alerta)
````

````python
# List comprehension também pode transformar os valores, não só filtrar
leituras_celsius = [72.0, 76.5, 91.0, 68.3]

# Convertendo Celsius para Fahrenheit: F = C * 9/5 + 32
leituras_fahrenheit = [celsius * 9 / 5 + 32 for celsius in leituras_celsius]

print("Celsius:   ", leituras_celsius)
print("Fahrenheit:", [round(f, 1) for f in leituras_fahrenheit])
````

## 3. Ordenando valores

`sorted()` retorna uma **nova lista** ordenada, sem alterar a original. Já `.sort()` (visto na Aula 05) altera a lista original.

````python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

crescente = sorted(leituras_temperatura)
decrescente = sorted(leituras_temperatura, reverse=True)

print("Original:  ", leituras_temperatura)
print("Crescente: ", crescente)
print("Decrescente:", decrescente)
````

### Ordenando estruturas mais complexas com `key`

Quando temos uma lista de dicionários, usamos o parâmetro `key` para indicar **por qual campo** ordenar.

````python
equipamentos = [
    {"nome": "Motor-01", "temperatura": 76.5},
    {"nome": "Bomba-03", "temperatura": 65.2},
    {"nome": "Compressor-02", "temperatura": 91.0},
]

# key=lambda recebe cada item da lista e devolve o valor usado para comparar
equipamentos_ordenados = sorted(equipamentos, key=lambda a: a["temperatura"], reverse=True)

for equipamento in equipamentos_ordenados:
    print(f"{equipamento['nome']}: {equipamento['temperatura']}°C")
````

## 4. Combinando filtragem e ordenação

É comum encadear as duas operações: primeiro filtrar, depois ordenar o resultado.

````python
equipamentos = [
    {"nome": "Motor-01", "temperatura": 76.5},
    {"nome": "Bomba-03", "temperatura": 65.2},
    {"nome": "Compressor-02", "temperatura": 91.0},
    {"nome": "Ventilador-01", "temperatura": 88.4},
]

limite_alerta = 80.0

em_alerta_ordenados = sorted(
    [equipamento for equipamento in equipamentos if equipamento["temperatura"] >= limite_alerta],
    key=lambda item: item["temperatura"],
    reverse=True,
)

for equipamento in em_alerta_ordenados:
    print(f"{equipamento['nome']}: {equipamento['temperatura']}°C")
````

## 5. Compreensão de dicionários (bônus)

O mesmo princípio das *list comprehensions* funciona para criar dicionários, usando `{chave: valor for ...}`.

````python
nomes = ["Motor-01", "Bomba-03", "Compressor-02"]
temperaturas = [76.5, 65.2, 91.0]

# zip() combina duas listas, par a par
mapa_temperaturas = {nome: temperatura for nome, temperatura in zip(nomes, temperaturas)}
print(mapa_temperaturas)
````

## 6. Resumo da aula

- Filtrar = percorrer uma sequência e manter apenas os itens que satisfazem uma condição.
- *List comprehensions* (`[expr for item in seq if cond]`) tornam filtragem e transformação mais compactas e legíveis.
- `sorted(sequencia, key=..., reverse=...)` ordena sem alterar a lista original.
- O mesmo padrão de compreensão também existe para dicionários.
