## 1. Por que criar funções

Nas aulas anteriores, repetimos algumas vezes a mesma lógica de classificar uma temperatura em `normal`/`alerta`/`crítico`. Uma **função** empacota um bloco de código sob um nome, para que possa ser **reutilizado** sempre que necessário, sem copiar e colar.

## 2. Definindo uma função com `def`

```python
def classificar_temperatura(temperatura):
    if temperatura < 75.0:
        status = "normal"
    elif temperatura < 90.0:
        status = "alerta"
    else:
        status = "crítico"
    return status


# Agora podemos reaproveitar a função quantas vezes quisermos
print(classificar_temperatura(72.0))
print(classificar_temperatura(82.0))
print(classificar_temperatura(95.0))
```

**Anatomia da função:**

- `def nome_da_funcao(parametros):` inicia a definição.
- O bloco indentado é o **corpo** da função, executado toda vez que ela é chamada.
- `return valor` devolve um resultado para quem chamou a função (e encerra a execução da função).
- `temperatura` é um **parâmetro**: um "espaço reservado" preenchido com o valor passado na chamada (o **argumento**).

## 3. Parâmetros com valor padrão

Podemos definir um valor padrão para um parâmetro, tornando-o opcional na hora de chamar a função.

```python
def classificar_temperatura(temperatura, limite_alerta=75.0, limite_critico=90.0):
    if temperatura < limite_alerta:
        return "normal"
    elif temperatura < limite_critico:
        return "alerta"
    else:
        return "crítico"


# Usando os limites padrão
print(classificar_temperatura(82.0))

# Sobrescrevendo os limites para um equipamento mais sensível
print(classificar_temperatura(82.0, limite_alerta=70.0, limite_critico=85.0))
```

## 4. Funções que recebem e retornam múltiplos valores

```python
def resumir_leituras(leituras):
    minimo = min(leituras)
    maximo = max(leituras)
    media = sum(leituras) / len(leituras)
    return minimo, maximo, media   # retorna uma tupla com três valores


leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]
minimo, maximo, media = resumir_leituras(leituras_temperatura)   # desempacotando a tupla

print(f"Mínimo: {minimo}°C | Máximo: {maximo}°C | Média: {media:.1f}°C")
```

## 5. Reaproveitando funções dentro de outras

Funções podem chamar outras funções — assim construímos lógicas mais completas a partir de blocos simples e testados.

```python
def classificar_temperatura(temperatura, limite_alerta=75.0, limite_critico=90.0):
    if temperatura < limite_alerta:
        return "normal"
    elif temperatura < limite_critico:
        return "alerta"
    else:
        return "crítico"


def classificar_lote(leituras):
    resultado = []
    for temperatura in leituras:
        status = classificar_temperatura(temperatura)   # reaproveitando a função anterior
        resultado.append((temperatura, status))
    return resultado


leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]
for temperatura, status in classificar_lote(leituras_temperatura):
    print(f"{temperatura}°C -> {status}")
```

## 6. Funções anônimas com `lambda`

Uma `lambda` é uma função **sem nome**, escrita em uma única linha, usada quando a lógica é curta e não vale a pena criar uma função completa com `def`. Vamos usar esse padrão na Aula 07, dentro de `sorted(..., key=lambda item: ...)`.

```python
# Função tradicional
def dobro(x):
    return x * 2

# Equivalente como lambda
dobro_lambda = lambda x: x * 2

print(dobro(5), dobro_lambda(5))
```

```python
# Uso típico de lambda: como argumento de outra função, sem precisar de um nome próprio
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

# map() aplica a lambda a cada elemento da lista
leituras_fahrenheit = list(map(lambda c: c * 9 / 5 + 32, leituras_temperatura))
print("Fahrenheit:", [round(f, 1) for f in leituras_fahrenheit])

# filter() mantém apenas os elementos em que a lambda retorna True
leituras_criticas = list(filter(lambda c: c >= 90.0, leituras_temperatura))
print("Críticas:", leituras_criticas)
```

**Quando usar `lambda` vs `def`:** prefira `lambda` apenas para lógicas de uma linha, geralmente passadas como argumento (`key=`, `map`, `filter`). Para qualquer lógica com mais de uma linha, ou que mereça um nome descritivo, use `def` — é mais legível e reutilizável.

## 7. Resumo da aula

- `def nome(parametros): ... return valor` cria uma função reutilizável.
- Parâmetros podem ter valores padrão, tornando-se opcionais.
- Uma função pode retornar múltiplos valores (como uma tupla) e pode chamar outras funções.
- `lambda parametros: expressao` cria uma função anônima para lógicas curtas, muito usada com `sorted`, `map` e `filter`.
