## 1. Listas (`list`)

Uma **lista** guarda uma sequência **ordenada** e **modificável** de valores. Já usamos listas na aula anterior — agora vamos explorar como manipulá-las.

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

print(leituras_temperatura)
print("Quantidade de leituras:", len(leituras_temperatura))
print("Primeira leitura:", leituras_temperatura[0])   # índices começam em 0
print("Última leitura:", leituras_temperatura[-1])    # índice negativo conta do fim
print("Três primeiras:", leituras_temperatura[0:3])   # fatiamento (slicing)
```

```python
# Listas são mutáveis: podemos alterar, adicionar e remover elementos
leituras_temperatura.append(79.4)          # adiciona ao final
print("Após append:", leituras_temperatura)

leituras_temperatura[0] = 73.0             # altera o valor na posição 0
print("Após alterar índice 0:", leituras_temperatura)

leituras_temperatura.remove(91.0)          # remove pelo valor
print("Após remove(91.0):", leituras_temperatura)

leituras_temperatura.sort()                # ordena a lista (em ordem crescente)
print("Ordenada:", leituras_temperatura)
```

## 2. Tuplas (`tuple`)

Uma **tupla** também é uma sequência ordenada, mas é **imutável**: uma vez criada, não pode ser alterada. Usamos tuplas para representar dados que não devem mudar, como coordenadas fixas de instalação de um sensor.

```python
coordenada_sensor = (-22.9068, -43.1729)   # (latitude, longitude) - não deve mudar

print(coordenada_sensor)
print("Latitude:", coordenada_sensor[0])
print("Longitude:", coordenada_sensor[1])

try:
    coordenada_sensor[0] = 0   # isso vai gerar erro: tuplas não podem ser alteradas
except TypeError as erro:
    print("Erro ao tentar alterar a tupla:", erro)
```

**Por que usar tupla em vez de lista?** Além de comunicar a intenção de que o dado não deve mudar, tuplas são levemente mais eficientes e podem ser usadas como chave de dicionário (o que listas não podem, como veremos adiante).

## 3. Conjuntos (`set`)

Um **conjunto** guarda valores **únicos**, sem ordem definida e sem repetição. É ideal para responder perguntas como "quais tipos de equipamento apareceram nas leituras?", eliminando duplicatas automaticamente.

```python
tipos_equipamento = ["Motor", "Bomba", "Motor", "Compressor", "Bomba", "Motor"]

tipos_unicos = set(tipos_equipamento)
print("Lista original:", tipos_equipamento)
print("Tipos únicos:", tipos_unicos)
print("Quantidade de tipos diferentes:", len(tipos_unicos))
```

```python
# Operações entre conjuntos
equipamentos_criticos = {"Motor-01", "Bomba-03", "Compressor-02"}
equipamentos_em_manutencao = {"Bomba-03", "Ventilador-01"}

print("União:", equipamentos_criticos | equipamentos_em_manutencao)
print("Interseção (críticos E em manutenção):", equipamentos_criticos & equipamentos_em_manutencao)
print("Diferença (críticos e NÃO em manutenção):", equipamentos_criticos - equipamentos_em_manutencao)
```

## 4. Dicionários (`dict`)

Um **dicionário** guarda pares **chave → valor**, o que é extremamente útil para representar um registro com vários atributos — por exemplo, todas as leituras de um único equipamento em um dado instante.

```python
leitura_sensor = {
    "equipamento": "Motor-01",
    "temperatura": 76.5,
    "pressao": 5.8,
    "vibracao": 2.1,
    "em_operacao": True,
}

print(leitura_sensor)
print("Temperatura:", leitura_sensor["temperatura"])   # acesso por chave
print("Pressão:", leitura_sensor.get("pressao"))        # .get() também acessa por chave
```

```python
# Alterando e adicionando chaves
leitura_sensor["temperatura"] = 77.0        # altera um valor existente
leitura_sensor["status"] = "alerta"         # adiciona uma nova chave
print(leitura_sensor)

# Percorrendo um dicionário
for chave, valor in leitura_sensor.items():
    print(f"{chave}: {valor}")
```

### Dicionário de listas: organizando várias leituras por equipamento

```python
leituras_por_equipamento = {
    "Motor-01": [72.0, 76.5, 79.4],
    "Bomba-03": [65.2, 66.8, 64.1],
}

leituras_por_equipamento["Compressor-02"] = [88.0, 90.5]   # novo equipamento

for equipamento, leituras in leituras_por_equipamento.items():
    media = sum(leituras) / len(leituras)
    print(f"{equipamento}: média = {media:.1f}°C (leituras: {leituras})")
```

## 5. Quando usar cada estrutura

| Estrutura | Ordenada | Modificável | Permite duplicatas | Uso típico |
|---|---|---|---|---|
| `list` | Sim | Sim | Sim | Sequência de leituras que pode crescer |
| `tuple` | Sim | Não | Sim | Dados fixos, como coordenadas |
| `set` | Não | Sim | Não | Valores únicos, checar pertencimento |
| `dict` | Sim (desde Python 3.7) | Sim | Chaves não, valores sim | Registro com vários atributos nomeados |

## 6. Resumo da aula

- `list`: sequência ordenada e mutável — a mais usada no dia a dia.
- `tuple`: sequência ordenada e imutável — para dados que não mudam.
- `set`: coleção de valores únicos, sem ordem — ótima para eliminar duplicatas.
- `dict`: pares chave-valor — ideal para representar registros com vários campos.
