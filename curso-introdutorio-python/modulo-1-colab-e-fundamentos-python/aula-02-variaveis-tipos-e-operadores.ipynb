## 1. Variáveis

Uma **variável** é um nome que guarda um valor na memória, para que possamos usá-lo depois. Em Python, criamos uma variável simplesmente atribuindo um valor a ela com o sinal `=`.

```python
# Criando variáveis para representar a leitura de um sensor industrial
equipamento = "Motor-01"
temperatura = 76.5
em_operacao = True

print(equipamento)
print(temperatura)
print(em_operacao)
```

**Regras para nomear variáveis:**

- Podem conter letras, números e `_`, mas não podem começar com número.
- São *case-sensitive*: `Temperatura` e `temperatura` são variáveis diferentes.
- Por convenção, usamos `snake_case` (palavras minúsculas separadas por `_`), como em `temperatura_maxima`.
- Evite nomes de funções e palavras reservadas do Python (`print`, `for`, `if`, etc.).

## 2. Tipos de dados

Python é uma linguagem de **tipagem dinâmica**: não precisamos declarar o tipo da variável, o próprio Python descobre a partir do valor atribuído. A função `type()` mostra o tipo de qualquer valor.

```python
equipamento = "Motor-01"      # str    -> texto (string)
temperatura = 76.5            # float  -> número com casas decimais
horas_operacao = 1200         # int    -> número inteiro
em_operacao = True            # bool   -> verdadeiro ou falso

print(equipamento, "->", type(equipamento))
print(temperatura, "->", type(temperatura))
print(horas_operacao, "->", type(horas_operacao))
print(em_operacao, "->", type(em_operacao))
```

### Conversão entre tipos (*type casting*)

Às vezes recebemos um dado em um tipo e precisamos converter para outro — por exemplo, uma leitura de sensor que chega como texto (`"76.5"`) e precisa virar número para podermos calcular com ela.

```python
leitura_bruta = "76.5"   # imagine que este valor veio de um arquivo de texto
print(type(leitura_bruta))

leitura_numerica = float(leitura_bruta)   # convertendo texto -> float
print(leitura_numerica, type(leitura_numerica))

# outras conversões comuns
print(int(3.9))          # float -> int (trunca, não arredonda!)
print(str(42))           # int -> str
print(bool(0), bool(1))  # int -> bool (0 é False, qualquer outro número é True)
```

## 3. Operadores aritméticos

Usados para fazer contas com números.

```python
temperatura_atual = 76.5
temperatura_limite = 80.0

soma = temperatura_atual + 2.0
diferenca = temperatura_limite - temperatura_atual
produto = temperatura_atual * 1.05        # simulando um aumento de 5%
divisao = temperatura_atual / 2           # divisão real (com casas decimais)
divisao_inteira = temperatura_atual // 2  # divisão inteira (descarta o resto)
resto = 77 % 5                            # resto da divisão (módulo)
potencia = 2 ** 3                         # potenciação

print("soma:", soma)
print("diferença:", diferenca)
print("produto:", produto)
print("divisão real:", divisao)
print("divisão inteira:", divisao_inteira)
print("resto:", resto)
print("potência:", potencia)
```

## 4. Operadores relacionais

Comparam dois valores e retornam um `bool` (`True` ou `False`). São a base das decisões que veremos na próxima aula.

```python
temperatura_atual = 76.5
temperatura_limite = 80.0

print(temperatura_atual == temperatura_limite)  # igual a
print(temperatura_atual != temperatura_limite)  # diferente de
print(temperatura_atual > temperatura_limite)   # maior que
print(temperatura_atual < temperatura_limite)   # menor que
print(temperatura_atual >= 76.5)                # maior ou igual a
print(temperatura_atual <= 76.5)                # menor ou igual a
```

## 5. Operadores lógicos

Combinam expressões booleanas usando `and` (e), `or` (ou) e `not` (negação). São essenciais para verificar **várias condições de um processo ao mesmo tempo**.

```python
temperatura = 76.5
pressao = 5.8

temperatura_ok = temperatura < 80.0
pressao_ok = pressao < 6.0

# 'and' -> só é True se AMBAS as condições forem verdadeiras
processo_normal = temperatura_ok and pressao_ok
print("Processo normal (and):", processo_normal)

# 'or' -> é True se PELO MENOS UMA condição for verdadeira
algum_alerta = (not temperatura_ok) or (not pressao_ok)
print("Existe algum alerta (or):", algum_alerta)

# 'not' -> inverte o valor booleano
print("Negação de temperatura_ok:", not temperatura_ok)
```

## 6. Formatando saídas com f-strings

As **f-strings** permitem inserir o valor de variáveis diretamente dentro de um texto, facilitando a criação de mensagens claras — muito útil para relatar o status de um equipamento.

```python
equipamento = "Motor-01"
temperatura = 76.5
limite = 80.0

mensagem = f"Equipamento {equipamento}: temperatura de {temperatura}°C (limite: {limite}°C)"
print(mensagem)

# f-strings também permitem controlar casas decimais
print(f"Margem até o limite: {limite - temperatura:.1f}°C")
```

## 7. Resumo da aula

- Variáveis guardam valores com `=`; o tipo é inferido automaticamente.
- Tipos principais: `int`, `float`, `str`, `bool`.
- Conversões de tipo são feitas com `int()`, `float()`, `str()`, `bool()`.
- Operadores aritméticos (`+ - * / // % **`), relacionais (`== != > < >= <=`) e lógicos (`and or not`) são a base de qualquer programa.
- f-strings (`f"texto {variavel}"`) formatam mensagens de forma legível.
