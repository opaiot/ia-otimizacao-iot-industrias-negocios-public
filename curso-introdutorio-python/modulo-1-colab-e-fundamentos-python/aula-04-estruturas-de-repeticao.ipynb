## 1. Por que repetir código

Imagine que precisamos verificar 100 leituras de temperatura, uma por uma, com o `if/elif/else` da aula anterior. Copiar e colar o mesmo bloco 100 vezes seria inviável. As **estruturas de repetição** (`for` e `while`) resolvem exatamente esse problema: executam o mesmo bloco de código várias vezes, automaticamente.

## 2. O laço `for`

O `for` percorre os elementos de uma sequência (uma lista, por exemplo) um de cada vez.

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

for temperatura in leituras_temperatura:
    print(f"Leitura: {temperatura}°C")
```

### `for` com `range()`

`range(n)` gera uma sequência de números de `0` até `n-1`. É útil quando precisamos repetir algo um número fixo de vezes, ou percorrer índices.

```python
for i in range(5):
    print("Verificação número", i)
```

## 3. Combinando `for` com `if`: automatizando a verificação

Aqui juntamos o que aprendemos nas Aulas 02 e 03 com o `for`, para classificar automaticamente **toda uma lista** de leituras.

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]
limite_alerta = 80.0
limite_critico = 90.0

for temperatura in leituras_temperatura:
    if temperatura < limite_alerta:
        status = "normal"
    elif temperatura < limite_critico:
        status = "alerta"
    else:
        status = "crítico"
    print(f"{temperatura}°C -> {status}")
```

### Contando ocorrências durante o laço

Um padrão muito comum: criar uma variável "contador" antes do laço e incrementá-la conforme uma condição é satisfeita.

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2, 95.5]
limite_critico = 90.0

total_criticas = 0

for temperatura in leituras_temperatura:
    if temperatura >= limite_critico:
        total_criticas = total_criticas + 1   # o mesmo que: total_criticas += 1

print(f"Total de leituras críticas: {total_criticas} de {len(leituras_temperatura)}")
```

## 4. `enumerate`: percorrendo valor e posição ao mesmo tempo

Quando precisamos saber **tanto o valor quanto sua posição** na lista (por exemplo, para identificar em qual leitura ocorreu o problema), usamos `enumerate`.

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

for indice, temperatura in enumerate(leituras_temperatura):
    if temperatura > 90.0:
        print(f"Leitura crítica encontrada na posição {indice}: {temperatura}°C")
```

## 5. O laço `while`

O `while` repete um bloco **enquanto uma condição for verdadeira**. É útil quando não sabemos de antemão quantas repetições serão necessárias — por exemplo, simular o resfriamento de um equipamento até atingir uma temperatura segura.

```python
temperatura = 95.0
limite_seguro = 80.0
minutos = 0

while temperatura > limite_seguro:
    temperatura -= 2.5   # simula o resfriamento de 2.5°C por minuto
    minutos += 1
    print(f"Minuto {minutos}: temperatura = {temperatura:.1f}°C")

print(f"Temperatura segura atingida após {minutos} minutos.")
```

**Cuidado:** se a condição do `while` nunca se tornar `False`, o laço roda para sempre (*loop infinito*). Sempre garanta que algo dentro do laço caminhe em direção ao fim da condição — no exemplo acima, é a linha `temperatura -= 2.5`.

## 6. `break` e `continue`

- `break` interrompe o laço imediatamente, mesmo que a condição ainda seja verdadeira.
- `continue` pula para a próxima repetição, sem executar o restante do bloco atual.

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

# break: para assim que encontrar a primeira leitura crítica
for temperatura in leituras_temperatura:
    if temperatura > 90.0:
        print(f"Parando verificação: leitura crítica encontrada ({temperatura}°C)")
        break
    print(f"{temperatura}°C está dentro do esperado, continuando...")
```

```python
leituras_temperatura = [72.0, -1.0, 91.0, 68.3, -1.0]  # -1.0 representa falha de leitura do sensor

# continue: ignora leituras inválidas e segue para as próximas
for temperatura in leituras_temperatura:
    if temperatura < 0:
        continue   # pula o restante do bloco para esta leitura
    print(f"Leitura válida: {temperatura}°C")
```

## 7. Resumo da aula

- `for` percorre sequências (listas, `range()`, etc.), um elemento por vez.
- `while` repete enquanto uma condição continuar `True` — cuidado com loops infinitos.
- `enumerate` fornece índice e valor simultaneamente.
- `break` interrompe o laço; `continue` pula para a próxima repetição.
- Combinar `for`/`while` com `if` é a base da automação de verificações em conjuntos de dados.
