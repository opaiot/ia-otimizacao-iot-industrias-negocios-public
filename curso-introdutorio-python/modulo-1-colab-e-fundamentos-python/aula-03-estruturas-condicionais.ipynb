## 1. Por que precisamos de condicionais

Até agora calculamos valores e comparamos números, mas um programa de verdade precisa **tomar decisões**: "se a temperatura passar de X, faça Y". É exatamente isso que as estruturas condicionais (`if`, `elif`, `else`) permitem.

## 2. A estrutura `if`

A sintaxe básica é:

```python
if condicao:
    # bloco executado somente se condicao for True
```

Repare que o Python usa **indentação** (espaços no início da linha) para definir o que pertence ao bloco do `if` — diferente de outras linguagens que usam `{ }`. O Colab indenta automaticamente após o `:`.

````python
temperatura = 85.0
limite = 80.0

if temperatura > limite:
    print("Alerta: temperatura acima do limite!")
````

## 3. `if` / `else`

O `else` define o que acontece quando a condição do `if` é `False`.

````python
temperatura = 72.0
limite = 80.0

if temperatura > limite:
    print("Alerta: temperatura acima do limite!")
else:
    print("Temperatura dentro do esperado.")
````

## 4. `if` / `elif` / `else`

Quando existem **mais de duas** possibilidades, usamos `elif` (contração de "else if") para encadear várias condições. É exatamente o cenário do OPAIOT: classificar um processo em **normal**, **alerta** ou **crítico** a partir da leitura de um sensor.

````python
temperatura = 91.0

if temperatura < 75.0:
    status = "normal"
elif temperatura < 90.0:
    status = "alerta"
else:
    status = "crítico"

print(f"Temperatura: {temperatura}°C -> status: {status}")
````

**Importante:** o Python avalia as condições **na ordem em que aparecem** e executa apenas o **primeiro** bloco cuja condição seja verdadeira, ignorando os demais. Por isso a ordem dos `elif` importa — nesse exemplo, se a temperatura for `91.0`, a primeira condição (`< 75.0`) é falsa, a segunda (`< 90.0`) também é falsa, então cai no `else`.

## 5. Combinando várias condições

Um processo real depende de mais de uma variável ao mesmo tempo. Podemos combinar condições com `and` e `or` diretamente dentro do `if`.

````python
temperatura = 78.0
pressao = 7.2
limite_temperatura = 80.0
limite_pressao = 6.5

if temperatura > limite_temperatura and pressao > limite_pressao:
    status = "crítico"          # os dois problemas ao mesmo tempo
elif temperatura > limite_temperatura or pressao > limite_pressao:
    status = "alerta"           # apenas um dos dois em excesso
else:
    status = "normal"

print(f"Temperatura: {temperatura}°C | Pressão: {pressao} bar -> status: {status}")
````

## 6. Condicionais aninhadas (`if` dentro de `if`)

Também é possível colocar um `if` dentro de outro, quando a segunda decisão só faz sentido depois que a primeira já foi avaliada.

````python
equipamento_ligado = True
temperatura = 95.0

if equipamento_ligado:
    if temperatura > 90.0:
        print("Equipamento ligado e em temperatura crítica: acionar parada de emergência!")
    else:
        print("Equipamento ligado e operando dentro da faixa.")
else:
    print("Equipamento desligado: nenhuma verificação de temperatura necessária.")
````

## 7. Expressão condicional em uma linha (operador ternário)

Para decisões simples de duas alternativas, o Python permite escrever o `if/else` em uma única linha — útil para atribuir um valor rapidamente.

````python
temperatura = 82.0
limite = 80.0

situacao = "acima do limite" if temperatura > limite else "dentro do limite"
print(situacao)
````

## 8. Resumo da aula

- `if` executa um bloco quando a condição é `True`.
- `elif` permite testar múltiplas condições em sequência; apenas o primeiro bloco verdadeiro é executado.
- `else` cobre todos os demais casos.
- Podemos combinar `and`/`or` dentro das condições para avaliar vários sensores ao mesmo tempo.
- Condicionais podem ser aninhadas, e decisões simples podem ser escritas em uma linha.
