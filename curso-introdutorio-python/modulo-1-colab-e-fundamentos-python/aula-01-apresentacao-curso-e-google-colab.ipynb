## 2. O que é o Google Colab

O [Google Colab](https://colab.research.google.com/) é um serviço gratuito do Google que permite escrever e executar código Python **direto no navegador**, sem precisar instalar nada no computador. Ele roda sobre um documento chamado **notebook** (extensão `.ipynb`), que mistura:

- **Células de texto (Markdown)** — como esta que você está lendo agora. Servem para explicações, títulos e anotações.
- **Células de código** — onde escrevemos e executamos Python de verdade.

O notebook fica salvo no seu Google Drive, pode ser compartilhado como um documento do Google e reaproveitado quantas vezes forem necessárias.

## 3. Executando uma célula

Para executar uma célula de código no Colab:

- Clique na célula e pressione **Shift + Enter** (executa e avança para a próxima célula); ou
- Pressione **Ctrl + Enter** (executa e permanece na mesma célula); ou
- Clique no botão de "play" (▶) que aparece à esquerda da célula.

O número entre colchetes à esquerda da célula (`[1]`, `[2]`, ...) indica a **ordem** em que as células foram executadas — não necessariamente a ordem em que aparecem no notebook. Por isso, se algo der errado, é uma boa prática usar o menu **Ambiente de execução → Executar tudo** para rodar o notebook do início.

## 5. Organização do notebook e boas práticas

- Cada aula deste curso é um **notebook separado**, salvo no Google Drive.
- Vá executando as células **na ordem em que aparecem**: cada célula pode depender do resultado das anteriores (por exemplo, uma variável criada em uma célula é usada em outra mais abaixo).
- Para salvar seu progresso, use **Arquivo → Salvar uma cópia no Drive** (ou `Ctrl+S`).
- Para baixar o notebook, use **Arquivo → Fazer download → .ipynb** ou **.py**.
- Se precisar de uma biblioteca que não vem instalada por padrão, o comando `!pip install nome_da_biblioteca` resolve (o `!` no início indica que é um comando de sistema, não Python).
