## 1. O que é uma biblioteca

Uma **biblioteca** (ou *library*/*package*) é um conjunto de código pronto, escrito por outras pessoas, que podemos reaproveitar em nossos programas em vez de reescrever do zero. Até aqui usamos apenas recursos nativos do Python; a partir de agora vamos importar bibliotecas especializadas em dados.

A primeira delas é o **NumPy** (*Numerical Python*), a base de praticamente todo o ecossistema de dados e Machine Learning em Python — inclusive o Pandas e o Scikit-learn, que veremos nos próximos módulos, são construídos sobre o NumPy.

No Google Colab, o NumPy já vem instalado. Só precisamos importá-lo — por convenção, sempre com o apelido `np`.

```python
import numpy as np

print("NumPy importado com sucesso, versão:", np.__version__)
```

## 2. Por que não usar listas comuns?

Listas Python são flexíveis, mas **lentas** para cálculos numéricos em grande volume, porque cada operação precisa passar por um laço `for` interno em Python puro. O NumPy resolve isso com o **array**, uma estrutura otimizada para guardar números do mesmo tipo e realizar cálculos de forma muito mais rápida (o NumPy delega o cálculo para código compilado em C).

## 3. Criando arrays

```python
leituras_temperatura = [72.0, 76.5, 91.0, 68.3, 85.2]

# convertendo uma lista Python em um array NumPy
array_temperaturas = np.array(leituras_temperatura)

print(array_temperaturas)
print(type(array_temperaturas))

print(array_temperaturas[array_temperaturas > 80])  # Filtrando temperaturas acima de 80
```

```python
# outras formas comuns de criar arrays
zeros = np.zeros(5)                 # array de 5 zeros
uns = np.ones(5)                    # array de 5 uns
sequencia = np.arange(0, 10, 2)     # de 0 a 10 (exclusivo), de 2 em 2 -> como range(), mas em array
espacados = np.linspace(0, 1, 5)    # 5 valores igualmente espaçados entre 0 e 1

print("zeros:    ", zeros)
print("uns:      ", uns)
print("arange:   ", sequencia)
print("linspace: ", espacados)
```

## 4. Atributos de um array

Todo array NumPy carrega informações sobre sua própria estrutura.

```python
array_temperaturas = np.array([72.0, 76.5, 91.0, 68.3, 85.2])

print("shape (formato):", array_temperaturas.shape)   # quantidade de elementos por dimensão
print("ndim (dimensões):", array_temperaturas.ndim)    # número de dimensões (1D, 2D, ...)
print("dtype (tipo dos dados):", array_temperaturas.dtype)  # tipo interno dos números
print("size (total de elementos):", array_temperaturas.size)
```

**Diferença importante em relação às listas:** um array NumPy guarda elementos de **um único tipo** (geralmente `float64` ou `int64`). Isso é o que permite as otimizações de desempenho — mas também significa que, se misturarmos tipos, o NumPy converte tudo para um tipo comum.

```python
mistura = np.array([1, 2.5, 3])   # int e float misturados
print(mistura, mistura.dtype)     # o NumPy converte tudo para float64
```

## 5. Indexação e fatiamento (*slicing*)

Funciona de forma parecida com listas Python, que já vimos na Aula 05.

```python
array_temperaturas = np.array([72.0, 76.5, 91.0, 68.3, 85.2, 60.1])

print("Primeiro elemento:", array_temperaturas[0])
print("Últimos dois:", array_temperaturas[-2:])
print("Do segundo ao quarto:", array_temperaturas[1:4])
```

### Indexação booleana: um recurso exclusivo (e poderoso) do NumPy

Podemos usar uma **condição** diretamente como índice, para selecionar apenas os elementos que a satisfazem — sem precisar escrever um `for`.

```python
array_temperaturas = np.array([72.0, 76.5, 91.0, 68.3, 85.2, 60.1])

mascara = array_temperaturas >= 80.0   # cria um array de True/False, o mesmo tamanho do original
print("Máscara booleana:", mascara)

leituras_em_alerta = array_temperaturas[mascara]   # usa a máscara para filtrar
print("Leituras em alerta:", leituras_em_alerta)

# a mesma ideia, em uma única linha:
print("Leituras em alerta (direto):", array_temperaturas[array_temperaturas >= 80.0])
```

## 6. Resumo da aula

- Bibliotecas trazem código pronto e reutilizável; importamos o NumPy com `import numpy as np`.
- O `array` do NumPy é otimizado para cálculos numéricos, muito mais rápido que listas Python para grandes volumes de dados.
- `np.array()`, `np.zeros()`, `np.ones()`, `np.arange()` e `np.linspace()` são formas comuns de criar arrays.
- `shape`, `ndim`, `dtype` e `size` descrevem a estrutura de um array.
- A indexação booleana (`array[condicao]`) permite filtrar dados sem escrever laços.
