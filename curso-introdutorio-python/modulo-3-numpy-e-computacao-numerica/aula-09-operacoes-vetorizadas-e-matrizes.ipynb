## 1. Operações vetorizadas

Na Aula 04, para converter uma lista inteira de Celsius para Fahrenheit, precisamos de um `for` (ou uma *list comprehension*). Com arrays NumPy, a mesma operação é escrita **diretamente sobre o array inteiro**, sem laço explícito — isso é chamado de **vetorização**.

```python
import numpy as np

leituras_celsius = np.array([72.0, 76.5, 91.0, 68.3, 85.2])

# a operação é aplicada a CADA elemento do array, automaticamente
leituras_fahrenheit = leituras_celsius * 9 / 5 + 32

print("Celsius:   ", leituras_celsius)
print("Fahrenheit:", np.round(leituras_fahrenheit, 1))
```

Além de ser muito mais concisa, a vetorização é **muito mais rápida** que o laço `for` equivalente, especialmente com milhares ou milhões de valores — o NumPy executa o cálculo em código compilado, evitando o custo de repetir a operação em Python puro a cada iteração.

## 2. Operações entre dois arrays

Quando os arrays têm o mesmo tamanho, as operações são feitas **elemento a elemento**.

```python
temperatura_motor = np.array([72.0, 76.5, 91.0, 68.3])
temperatura_ambiente = np.array([25.0, 26.0, 24.5, 25.5])

diferenca = temperatura_motor - temperatura_ambiente
print("Diferença motor - ambiente:", diferenca)

media_dos_dois = (temperatura_motor + temperatura_ambiente) / 2
print("Média:", media_dos_dois)
```

## 3. *Broadcasting*

O NumPy também permite operar um array com um único número (*escalar*), aplicando a operação a todos os elementos — isso é chamado de ***broadcasting***. Já fizemos isso sem perceber: `leituras_celsius * 9 / 5 + 32` fez *broadcasting* do `9`, `5` e `32` para todos os elementos do array.

```python
leituras = np.array([72.0, 76.5, 91.0, 68.3])
ajuste_calibracao = 1.5   # escalar: um único número

leituras_calibradas = leituras + ajuste_calibracao   # somado a TODOS os elementos
print("Leituras calibradas:", leituras_calibradas)
```

## 4. Matrizes: arrays de duas dimensões

Uma **matriz** é um array 2D: linhas e colunas. É útil para organizar, por exemplo, várias leituras (colunas) de vários sensores (linhas).

```python
# cada linha representa um equipamento; cada coluna, uma leitura ao longo do tempo
leituras_por_equipamento = np.array([
    [72.0, 74.5, 76.0, 75.2],   # Motor-01
    [65.2, 66.8, 64.1, 65.9],   # Bomba-03
    [88.0, 90.5, 91.2, 89.7],   # Compressor-02
])

print(leituras_por_equipamento)
print("Formato (linhas, colunas):", leituras_por_equipamento.shape)
```

## 5. Indexação em matrizes

Usamos `matriz[linha, coluna]`. Também podemos selecionar linhas ou colunas inteiras usando `:`.

```python
leituras_por_equipamento = np.array([
    [72.0, 74.5, 76.0, 75.2],   # Motor-01     (linha 0)
    [65.2, 66.8, 64.1, 65.9],   # Bomba-03     (linha 1)
    [88.0, 90.5, 91.2, 89.7],   # Compressor-02 (linha 2)
])

print("Elemento na linha 0, coluna 2:", leituras_por_equipamento[0, 2])
print("Todas as leituras do Compressor-02 (linha 2):", leituras_por_equipamento[2, :])
print("Todas as leituras no instante 0 (coluna 0):", leituras_por_equipamento[:, 0])
print("Duas primeiras linhas, duas últimas colunas:")
print(leituras_por_equipamento[0:2, 2:4])
```

## 6. Operações ao longo de linhas e colunas (`axis`)

Muitas funções do NumPy aceitam o parâmetro `axis` para indicar a **direção** do cálculo:

- `axis=0` → percorre as **linhas**, ou seja, calcula "por coluna" (por instante de tempo).
- `axis=1` → percorre as **colunas**, ou seja, calcula "por linha" (por equipamento).

```python
leituras_por_equipamento = np.array([
    [72.0, 74.5, 76.0, 75.2],   # Motor-01
    [65.2, 66.8, 64.1, 65.9],   # Bomba-03
    [88.0, 90.5, 91.2, 89.7],   # Compressor-02
])

media_por_equipamento = leituras_por_equipamento.mean(axis=1)   # uma média por linha
media_por_instante = leituras_por_equipamento.mean(axis=0)      # uma média por coluna

print("Média de temperatura por equipamento:", np.round(media_por_equipamento, 1))
print("Média de temperatura por instante:   ", np.round(media_por_instante, 1))
```

## 7. Transposição e reshape

`transpose()` (ou `.T`) troca linhas por colunas. `reshape()` reorganiza os elementos em um novo formato, mantendo a mesma quantidade total.

```python
leituras_por_equipamento = np.array([
    [72.0, 74.5, 76.0, 75.2],
    [65.2, 66.8, 64.1, 65.9],
    [88.0, 90.5, 91.2, 89.7],
])

print("Transposta (instantes nas linhas, equipamentos nas colunas):")
print(leituras_por_equipamento.T)

sequencia = np.arange(12)
print("\nSequência original:", sequencia)
print("Reorganizada em 3 linhas x 4 colunas:")
print(sequencia.reshape(3, 4))
```

## 8. Resumo da aula

- Operações vetorizadas aplicam um cálculo a todo o array de uma vez, sem laços explícitos — mais rápido e mais legível.
- *Broadcasting* permite operar um array com um escalar (ou arrays de formatos compatíveis).
- Matrizes são arrays 2D; acessamos elementos com `matriz[linha, coluna]`.
- `axis=0` opera por coluna, `axis=1` opera por linha.
- `.T` transpõe uma matriz; `.reshape()` reorganiza os elementos em outro formato.
